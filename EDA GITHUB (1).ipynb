```python
import requests
import pandas as pd
import time
from tqdm import tqdm
```

```python
categories = {

"Data Science":[
"data-science",
"machine-learning",
"deep-learning",
"artificial-intelligence",
"nlp",
"computer-vision",
"generative-ai",
"llm"
],

"Web Development":[
"html",
"css",
"javascript",
"react",
"nodejs"
],

"Cybersecurity":[
"cybersecurity",
"malware",
"pentesting",
"security-tools"
],

"Full Stack":[
"mern",
"mean",
"mevn",
"full-stack",
"web-development",
"web-app",
"crud",
"rest-api",
"graphql"
],

"Database":[
"mysql",
"postgresql",
"mongodb",
"sqlite",
"redis",
"firebase",
"supabase",
"prisma"
],

"JavaScript Frameworks":[
"react",
"nextjs",
"vue",
"nuxt",
"angular",
"svelte",
"solidjs",
"astro"
],

"UI UX":[
"ui",
"ux",
"dashboard",
"admin-panel",
"landing-page",
"portfolio",
"ecommerce",
"blog"
],

"API":[
"rest-api",
"graphql",
"swagger",
"openapi",
"api",
"postman"
]
}
```

```python
dataset=[]
```

```python
for category, topics in categories.items():

    print("=" * 60)
    print(category)
    print("=" * 60)

    for topic in tqdm(topics):

        url = f"https://api.github.com/search/repositories?q={topic}&sort=stars&order=desc&per_page=100"

        response = requests.get(url)

        if response.status_code != 200:
            print(f"Error {response.status_code} for {topic}")
            continue

        data = response.json()

        if "items" not in data:
            print(f"No repositories found for {topic}")
            continue

        for repo in data["items"]:

            dataset.append({

                "Category": category,
                "Topic": topic,
                "Repository": repo["name"],
                "Owner": repo["owner"]["login"],
                "Repository URL": repo["html_url"],
                "Language": repo["language"],
                "Stars": repo["stargazers_count"],
                "Forks": repo["forks_count"],
                "Watchers": repo["watchers_count"],
                "Open Issues": repo["open_issues_count"],
                "Created At": repo["created_at"],
                "Updated At": repo["updated_at"]

            })

        time.sleep(2)
```

```python
df=pd.DataFrame(dataset)
```

```python
#2. Hypothesis Testing
from scipy import stats
```

```python
#A. Independent T-Test
python_stars = df[df["Language"]=="Python"]["Stars"]
java_stars   = df[df["Language"]=="Java"]["Stars"]

t_stat, p_value = stats.ttest_ind(
    python_stars,
    java_stars,
    equal_var=False
)

print(t_stat)
print(p_value)


#Decision
if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null")
```

```python
#B.Chi-Square Test
table = pd.crosstab(df["Category"], df["Language"])

chi2, p, dof, expected = stats.chi2_contingency(table)

print(chi2)
print(p)
```

```python
#Anova 
groups = [
    group["Stars"].values
    for _, group in df.groupby("Category")
]

f_stat, p = stats.f_oneway(*groups)

print(f_stat)
print(p)
```

```python
#3.Confidence Interval Analysis
import numpy as np
from scipy.stats import norm

mean = df["Stars"].mean()
std = df["Stars"].std()
n = len(df)

z = norm.ppf(0.975)
margin = z * (std / np.sqrt(n))

lower = mean - margin
upper = mean + margin

print(lower, upper)
```

```python
#4.Feature Engineering
#A. Popularity Category
df["Popularity Category"] = pd.cut(
    df["Stars"],
    bins=[0,100,1000,10000,1000000],
    labels=["Low","Medium","High","Very High"]
)
```

```python
#B. Repository Age
df["Created At"] = pd.to_datetime(df["Created At"],utc=True)

df["Repository Age"] = (
    pd.Timestamp.now(tz="UTC")-df["Created At"]
).dt.days // 365
```

```python
#C. Active Repository Flag
df["Active Repo"] = (
    (
        pd.Timestamp.today() - df["Updated At"]
    ).dt.days < 180
)
```

```python
#D.Engagement Score
df["Engagement Score"] = (
    df["Stars"] +
    df["Forks"]*2 +
    df["Watchers"]
)
```
